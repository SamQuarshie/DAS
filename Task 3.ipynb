9.	Using the clients dictionary from Task 7, write code using len(), sum(), max(), and min() to print:

    a.	The total number of clients

    b.	The total loan amount given out across all clients

    c.	The largest single loan amount
    
    d.	The smallest single loan amount


```python
# Task 9
all_loans = [details["loan_amount"] for details in clients.values()]
print("Total clients:", len(clients))
print("Total loans given:", sum(all_loans))
print("Largest loan:", max(all_loans))
print("Smallest loan:", min(all_loans))
```
